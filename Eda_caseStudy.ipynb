```python
#distribution of age
counts,edges,bars= plt.hist(data['Age'],bins=15,color='red')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title("Age Distribution")
plt.show()
```

```python
plt.bar(data['Survived'], data['Sex'])
plt.xlabel("Survived Rate")
plt.ylabel("Gender")
plt.title("Gender Survived Distribution")
plt.show()
```

```python
index=data['Survived'].value_counts().keys().tolist()
print(index)
labels=data['Survived'].value_counts().tolist()
print(labels)
```

```python
plt.pie(x=labels,
labels=['Dead','Alive'],
colors=['red','yellow'],
autopct="%.1f%%")
plt.title("Gender Distribution")
plt.legend()
plt.show()
```

```python
plt.scatter(x=data['Age'],
y=data['Fare'],)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```

```python
import plotly as py    
```

```python
import plotly.graph_objects as go
```

```python
trace = go.Histogram(x =data['Age'],
nbinsx=20,
marker={'color':'black'})
#layout define for the graph
layout =go.Layout(title="Age Distribution")
#we combine the trace with layout to display the figure
fig = go.Figure(data=trace,layout = layout)
fig.show()
```

```python
trace = go.Scatter(x=data['Fare'],
               y=data['Age'],
               mode='markers',
               marker={'color':'red'})
layout = go.Layout(title="age vs Pclass distribution")
fig = go.Figure(data=trace,layout = layout)
fig.show()
.






























































































```
